# IterAlign - Inference Demo

This notebook demonstrates how to run inference with a model trained using IterAlign.

```python
import sys
sys.path.append('..')

import yaml
from src.model import ensure_model_downloaded, load_model_and_tokenizer, generate_responses
```

```python
with open('../config.yaml') as f:
    config = yaml.safe_load(f)

model_cfg = config['model']
local_dir = ensure_model_downloaded(model_cfg['repo'], model_cfg['local_dir'])
model, tokenizer = load_model_and_tokenizer(local_dir)
```

## Test with example prompts

```python
prompts = [
    "What are the benefits of regular exercise?",
    "Explain how encryption works.",
    "Write a short poem about AI.",
]

responses = generate_responses(model, tokenizer, prompts, max_new_tokens=150)

for p, r in zip(prompts, responses):
    print(f"Q: {p}\nA: {r}\n")
    print("---")
```
